## 1. System Design & Implementation
The Person class is a base class that models an entity with two attributes: name and age. These attributes are defined and initialized within the constructor (__init__), which is automatically invoked each time an object is instantiated. Input validation is performed within the constructor to ensure that the age attribute is realistic (age cannot be negative), thereby preserving data integrity and preventing the creation of invalid object instances. The use of type annotations (str and int) enforces expected data types and improves code readability and maintainability. The class exposes a public method (get_details), which returns a formatted string representation of the state of the object, providing controlled access to its internal data.

```python
class Person:
    """
    Represents a generic Person. Will be used as base for persons in the Hotel.

    Attributes: 
        name (str): The name of the person.
        age (int): The age of the person.
    """
    
    def __init__(self, name: str, age: int) -> None:
        """
        Initialize a Person object.

        Args:
            name (str): The name of the person.
            age (int): The age of the person.

        Returns:
            None
        """
        if age < 0:
            raise ValueError("Age must be a positive integer.")
        self.name = name
        self.age = age

    def get_details(self) -> None:
        """
        Print the basic details of the Person.

        Args: 
            None
        
        Returns:
            None
        """
        print(f"Name: {self.name}, Age: {self.age}")
```

The validation logic implemented in the Person class constructor is inherited by both the Guest and Staff subclasses that follow, through the use of the super() function. As a result, all instances of these subclasses are also subject to the same constraints, ensuring consistency and data integrity across the system.

The subclass Guest extends the functionality of the base Person class. It also has attributes and behavior relevant to the hotel management context. It reuses the core properties of a Person (name, age) hrough inheritance, while also introducing the additional attibutes room_number and nights, which represent the accommodation details of a guest. Moreover, the class defines the method calculate_bill, which encapsulates the logic required to compute the total cost of a guest’s stay based on a given nightly rate. This method demonstrates how behavior can be specialized within a subclass to reflect context-specific operations.

```python
class Guest(Person):
    """
    Represents a Guest, which is a type of Person.

    Attributes:
        room_number (int | None): The room number assigned to the guest. Defaults to None if not provided.
        nights (int | None): The number of nights the guest will stay. Defaults to None if not provided.

    A guest has the ability to calculate the total bill based on the number of nights and a given room rate.
    """
    
    def __init__(self, name: str, age: int, room_number: int | None = None, nights: int | None = None) -> None:
        """
        Initialize a Guest object.

        Args:
            name (str): The name of the guest.
            age (int): The age of the guest.
            room_number (int | None): The room number assigned to the guest. Defaults to None if not provided.
            nights (int | None): The number of nights the guest will stay. Defaults to None if not provided.

        Returns:
            None
        """
        super().__init__(name, age)
        self.room_number = room_number
        self.nights = nights

    def calculate_bill(self, rate_per_night: float) -> float:
        """
        Calculate the total bill for the guest's stay based on the number of nights and the rate per night.
        
        Args:
            rate_per_night (float): The cost of one night in the hotel.
        
        Returns:
            float: The total bill for the guest's stay.
        """
        if self.nights is None:
            print("Number of nights is not set for this guest. Total bill will be 0.0")
            return 0.0
        return self.nights * rate_per_night
```

The subclass Staff extends the functionality of the base Person class in the same way that is described above for the Guest class. It reuses the core properties of a Person (name, age) through inheritance, and introduces the additional attributes position and salary, which describe the role and compensation of a staff member. Additionally, the class defines the method give_raise, which encapsulates the logic required to increase the staff member’s salary by a specified amount.

```python
class Staff(Person):
    """
    Represents a Staff member, which is a type of Person.

    Attributes:
        position (str): The job position of the staff member.
        salary (float): The salary of the staff member.

    Staff can increase the salary by a specified amount.
    """
    
    def __init__(self, name: str, age: int, position: str, salary: float) -> None:
        """
        Initialize a new instance of the Staff class.
        
        Args:
            name (str): The name of the staff member.
            age (int): The age of the staff member.
            position (str): The job position of the staff member.
            salary (float): The salary of the staff member.

        Returns:
            None
        """
        super().__init__(name, age)
        self.position = position
        self.salary = salary

    def give_raise(self, amount: float) -> None:
        """
        Increase the salary of the staff member by a specified amount.
        
        Args:
            amount (float): The amount to increase the salary by.

        Returns: 
            None
        """
        self.salary += amount
```

It is important to note that aforementioned monetary values such as salary and billing amounts are maintained as floating-point numbers for computational accuracy and will be formatted to two decimal places during output, for consistency in financial representation.

The Hotel class represents the central entity of the system, encapsulating the core functionality required for managing guests and staff within a hotel environment. It's attributes are a string name, representing the name of the hotel, and two lists, guests and staff, which store instances of the Guest and Staff classes respectively. These lists are initialized in a controlled manner to ensure proper handling of default values. For example, it is safe to initialize an "empty" Hotel, one that has just opened and does not have any Guests or Staff yet.

The class defines several methods that implement the main operations of the system. The check_in and check_out methods manage the addition and removal of guests from the hotel, while the hire_staff method is responsible for adding new staff members. Additionally, the methods list_guests and list_staff provide a way to retrieve structured information about all current guests and staff members by utilizing the get_details method inherited from the Person class.

Overall, the Hotel class demonstrates how multiple objects can be aggregated and managed within a single structure, highlighting the concept of composition, while also organizing the system’s functionality into clear and reusable operations.

```python
class Hotel:
    """
    Represents a Hotel that manages guests and staff.

    Attributes:
        name (str): The name of the hotel.
        guests (list, optional): A list of Guest instances representing the guests currently checked in the hotel. Defaults to None.
        staff (list, optional): A list of Staff instances representing the staff members working in the hotel. Defaults to None.
    
    It can check in and out guests, hire staff, and list guests and staff.
    """
    
    def __init__(self, name: str, guests: list = None, staff: list = None) -> None:
        """
        Initialize an instance of a Hotel.

        Args:
            name (str): The name of the hotel.
            guests (list, optional): A list of Guest instances representing the guests currently checked in the hotel. Defaults to None.
            staff (list, optional): A list of Staff instances representing the staff members working in the hotel. Defaults to None.

        Returns: 
            None
        """
        self.name = name
        self.guests = guests if guests is not None else []
        self.staff = staff if staff is not None else []

    def check_in(self, guest: Guest) -> None:
        """
        Add a guest to the hotel's guest list.

        Args:
            guest (Guest): An instance of the Guest class representing the guest to be checked in

        Returns: 
            None
        """
        self.guests.append(guest)
    
    def check_out(self, guest: Guest) -> None:
        """
        Remove a guest from the hotel's guest list.

        Args:
            guest (Guest): An instance of the Guest class representing the guest to be checked out

        Returns: 
            None
        """
        self.guests.remove(guest)

    def hire_staff(self, staff_member: Staff) -> None:
        """
        Add a staff member to the hotel's staff list.

        Args:
            staff_member (Staff): An instance of the Staff class representing the staff member to be hired

        Returns: 
            None
        """
        self.staff.append(staff_member)
    
    def list_guests(self) -> None:
        """
        Print the details of all guests currently checked in the hotel.

        Args:
            None

        Returns: 
            None
        """
        print("Guests currently checked in:")
        for guest in self.guests:
            guest.get_details()

    def list_staff(self) -> None:
        """
        Print the details of all staff members in the hotel.
        
        Args:
            None

        Returns: 
            None
        """
        print("Current staff members:")
        for staff_member in self.staff:
            staff_member.get_details()
```
